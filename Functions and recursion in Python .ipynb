#### FUNCTIONS IN PYTHONS

```python
def cal_avg(a,b,c):               #calculate the average of three function using function
    sum = a+b+c
    avg = sum/3
    print(avg)
cal_avg(1,2,3)
cal_avg(98,34,23)
```

```python
# programm to find the factorial of a number

def cal_fact(num):
    fact=1
    for i in range (1,num+1):
        fact=fact*i
    print(fact)
cal_fact(5)
```

```python
# program to convert USD in INR

def converter(usd_val):
    inr_val = usd_val * 83
    print(usd_val,"USD=",inr_val,"INR")
    
converter(73)    
```

### RECURSION

```python
# recursion function
def show(n):
    if(n==0):
        return
    print(n)
    show(n-1)

show(7)
```

```python
# factorial using recursion
def fact(n):
    if(n==1 or n==0):
        return 1
    return fact(n-1)*n
print(fact(6))
```

```python
# program to write the sum of first n number
def sum(n):
    if(n==0):
        return 0
    return sum(n-1)+n
print(sum(10))
```

```python
# program to print all element in a list.
cities = ["mumbai","delhi","noida","gujrat","pure"]

def print_list(list,idx=0):
    if(idx == len(list)):
        return
    print(list[idx])
    print_list(list,idx+1)
print_list(cities)
```
